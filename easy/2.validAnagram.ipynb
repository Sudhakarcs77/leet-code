```python
# method 1  , 2 maps, O(s+t)

def validAnagram(s,t):
    if len(s) != len(t):
        return False
    countS, countT = {},{}
    for char in s:
        countS[char] = countS.get(char, 0) + 1
    
    for char in t:
        countT[char] = countT.get(char, 0) + 1
    
    for c in countS:
        if countS[c] != countT.get(c, 0):
            return False
    return True

#test

s = "anagram"
t = "nagaram"
print(validAnagram(s, t)) # True
```

```python
# method 2

def validAnagram(s, t):
    if len(s) != len(t):
        return False
    return sorted(s) == sorted(t)

s = "anagram"
t = "nagaram"
print(validAnagram(s, t)) # True
```
