```python
#1. Contains Duplicates

# method 1
def is_duplicate(n):

    for i in range(len(n)):
        for j in range(i+1, len(n)):
            if n[i] == n[j]:
                return True
    return False

# Test Cases
print(is_duplicate([1, 2, 3, 4, 5]))  # False
print(is_duplicate([1, 2, 3, 4, 5, 5]))  # True
```

```python
# method 2

def is_duplicate(n):
    n.sort() #

    for i in range(1,len(n)):
        if n[i]-1 == n[i]:
            return True
    return False
    
print(is_duplicate([1, 2, 3, 4, 5]))  # False
print(is_duplicate([1, 2, 3, 4, 5, 5]))  # True
```

```python
# method 3 using map - O(n), O(n)

def is_duplicate(n):
    map = {}
    for i in range(len(n)):
        if n[i] in map.keys():
            return True
        else:
            map[n[i]] = True
        
    return False

print(is_duplicate([1, 2, 3, 4, 5]))  # False
print(is_duplicate([1, 2, 3, 4, 5, 5]))  # True
```
